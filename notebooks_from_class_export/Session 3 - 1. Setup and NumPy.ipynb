## ­ƒôè NumPy ÔÇö Statistical Functions <a class="anchor" id="numpy"></a>

**NumPy** is the numerical computing library at the core of Python data science. Every column of numbers in a Pandas DataFrame is a NumPy array under the hood ÔÇö so when you compute the mean of a salary or views column later today, you are using NumPy.

For this session, the functions you need are the statistical ones:

### ­ƒôê 5. Statistical Functions

These are the functions you will use most often when analysing job postings data.

| Function | Code | Description |
|---|---|---|
| Mean | `arr.mean()` | Average value |
| Std deviation | `arr.std()` | Spread of the data |
| Min / Max | `arr.min()` / `arr.max()` | Lowest / highest value |
| Median | `np.median(arr)` | Middle value |
| Sum | `arr.sum()` | Total |

```python
temps = np.array([19.5, 20.1, 18.9, 21.3, 22.0, 19.8, 20.6])

print(f"Mean:    {temps.mean():.2f}")
```

```python
tmp2 = np.array([1, 2, 3])
tmp2.mean()
```

```python
print(f"Std:     {temps.std():.2f}")
```

```python
print(f"Min:     {temps.min():.2f}")
```

```python
print(f"Max:     {temps.max():.2f}")
```

```python
print(f"Median:  {np.median(temps):.2f}")
```

### ­ƒÅó 6. Real-World Context: Job Postings Data as Arrays

In the dataset we will work with throughout this course, each numerical column is a NumPy array. When Pandas loads data from the database, every numerical column is stored as a NumPy array internally.

For example, a sample of view counts from the LinkedIn postings dataset:

```python
# A sample of view counts from the LinkedIn job postings dataset
views = np.array([45, 52, 48, 61, 55, 49, 58, 120, 134, 118])

print(f"Views mean:  {views.mean():.1f}")
print(f"Views std:   {views.std():.1f}")
print(f"Views max:   {views.max():.1f}")
```

```python
# Is any posting more than 1 standard deviations above the mean?
threshold = views.mean() + 1 * views.std()
threshold
```

```python
views[views > threshold]
```
