## ­ƒôè GroupBy ÔÇö Aggregation by Category <a class="anchor" id="groupby"></a>

### ­ƒº® 1. `value_counts()` and `groupby()`

The quickest way to count by category is `value_counts()`. For other aggregations (mean, median, sum), use `groupby()`.

````python
# Count postings by work type
df_postings['work_type'].value_counts()
````

````python
# Count postings by experience level
df_postings['formatted_experience_level'].value_counts()
````

```sql
SELECT 
    work_type, 
    ROUND(AVG(views), 1) as avg_views
FROM postings
GROUP BY work_type
ORDER BY avg_views DESC
```

````python
# Average views by work type
df_postings.groupby('work_type')[['views']].mean().round(1).sort_values(by='views')
````

## ­ƒôè Multiple Aggregations <a class="anchor" id="multi"></a>

### ­ƒº® 3. `.agg()` ÔÇö Several Statistics at Once

Instead of computing mean, median, and count separately, pass them all to `.agg()`.

```sql
SELECT col1, sum(col3), count(col2), max(col4), max(col2)
FROM table
WHERE col2 = 1
GROUP BY col1
```

````python
# Views statistics by experience level
(
df_postings
    .groupby('formatted_experience_level')
    ['views'].agg(['count','mean','median','std']).round(1)
)
````

````python
# Salary statistics by work type
# Note: count will be much lower than total postings ÔÇö groupby ignores NaN by default
(
    df_postings
        .groupby('work_type')
        ['normalized_salary'].agg(['count','mean','median']).round(0)
)
````

````python
# Compare: total postings vs postings with salary data per work type
total = df_postings.groupby('work_type').size().rename('total')
with_salary = df_postings.groupby('work_type')['normalized_salary'].count().rename('with_salary')
pd.concat([total, with_salary], axis=1)
````

## ­ƒöì GroupBy with Filtering <a class="anchor" id="filter"></a>

### ­ƒº® 4. Filter Before Aggregating

When a column has many nulls, always filter first ÔÇö otherwise your counts will be misleading.

````python

df_salary = df_postings.dropna(subset=['normalized_salary'])

df_salary_grouped = (
    df_salary
    .groupby('formatted_experience_level')
        ['normalized_salary'].median()
    .sort_values(ascending=False)
    .round(0)
)
````

````python
df_salary_grouped
````

## ­ƒôï Multi-Column GroupBy <a class="anchor" id="multicol"></a>

### ­ƒº® 5. Group by Two Columns

Pass a list of column names to `groupby()` to create a cross-tabulation.

```sql
SELECT col1, col2, sum(col3)
FROM table
WHERE col2 = 1
GROUP BY col1, col2
```

````python
# Count postings by work type AND experience level
(
    df_postings
    .groupby(['work_type', 'formatted_experience_level'])
    .size() #count rows
        #remove the 'work_type', 'formatted_experience_level' from index 
    .reset_index(name='count') 
    .sort_values('count', ascending=False)
    .head(15)
)
````

## ­ƒôë Rolling Windows <a class="anchor" id="rolling"></a>

### ­ƒº® 6. Smoothing a Series with `.rolling()`

A **rolling window** computes a statistic over a sliding window of N consecutive rows. It is used to smooth noisy data and reveal underlying trends.

```python
df['col'].rolling(window=7).mean()   # 7-row rolling average
```

The first `window - 1` rows will be `NaN` ÔÇö there are not enough preceding values yet.

````python
df_postings.head(5)
````

````python
df_postings.set_index('listed_time').resample('1D')
````

````python
# Daily posting count ÔÇö resample listed_time to daily
daily_counts = (
    df_postings
    .set_index('listed_time')
    .resample('1D')
    .size()
    .rename('postings')
)
daily_counts
````

````python
# 3-day rolling average ÔÇö smooths day-to-day noise
daily_counts_smooth = daily_counts.rolling(window=3).mean().round(1)

pd.DataFrame({
    'raw': daily_counts,
    'rolling_3d': daily_counts_smooth
})
````

> **Why rolling matters:** Even with only 4 weeks of data, rolling averages remove single-day spikes and make the underlying posting rate clearer. The same technique applies to any sequential data ÔÇö stock prices, sensor readings, website traffic.

## Ô£à Summary

| Concept | Code | Use |
|---|---|---|
| Count by category | `df['col'].value_counts()` | Quick frequency count |
| GroupBy + single stat | `df.groupby('col')['val'].mean()` | One aggregation |
| GroupBy + multiple stats | `.agg(['count','mean','median'])` | Several at once |
| Filter before groupby | `df.dropna(subset=['col'])` | Avoid misleading NaN counts |
| Multi-column groupby | `df.groupby(['a','b']).size()` | Cross-tabulation |
| Rolling window | `.rolling(window=N).mean()` | Smooth noisy sequential data |
