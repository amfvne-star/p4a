## ­ƒÄ¿ Matplotlib ÔÇö Line and Bar Charts <a class="anchor" id="matplotlib"></a>

### ­ƒº® 1. Line Chart ÔÇö Trend Over Time

Line charts are ideal for showing how a value changes over time. Here we resample postings to a daily count to see the posting rate across the dataset's date range.

```python
# Daily posting count over time
daily_counts = (
    df_postings
    .set_index('listed_time')
    .resample('1D')
    .size()
)
```

```python
fig, ax = plt.subplots(figsize=(14, 4))

daily_counts.plot(ax=ax, color='steelblue', linewidth=1.5)

ax.set_title('Daily Job Postings Over Time', fontsize=14)
ax.set_xlabel('Date')
ax.set_ylabel('Number of Postings')
plt.tight_layout()
plt.show()
```

```python
# Two metrics over time ÔÇö subplots with shared x-axis
daily_views = (
    df_postings
    .set_index('listed_time')
    .resample('1D')['views']
    .mean()
)
```

```python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 6), sharex=True)

daily_counts.plot(ax=ax1, color='steelblue', linewidth=1.5)
ax1.set_title('Daily Posting Count', fontsize=12)
ax1.set_ylabel('Postings')

daily_views.plot(ax=ax2, color='coral', linewidth=1.5)
ax2.set_title('Daily Average Views', fontsize=12)
ax2.set_ylabel('Avg Views')
ax2.set_xlabel('Date')

plt.tight_layout()
plt.show()
```

### ­ƒº® 2. Bar Chart ÔÇö Compare Categories

Bar charts are ideal for comparing a single metric across a small number of categories.

```python
work_type_counts = df_postings['formatted_work_type'].value_counts()
work_type_counts
```

```python
# Posting count by work type
fig, ax = plt.subplots(figsize=(10, 5))

work_type_counts.plot.bar(ax=ax, color='steelblue', edgecolor='white')

ax.set_title('Job Postings by Work Type', fontsize=14)
ax.set_xlabel('Work Type')
ax.set_ylabel('Number of Postings')
ax.tick_params(axis='x', rotation=30)
plt.tight_layout()
plt.show()
```

### ­ƒº® 3. Sorted Bar Chart and Horizontal Bar

Sorting bars makes comparisons easier. Horizontal bars work well when category labels are long.

```python
# Median salary by experience level ÔÇö sorted ascending
exp_salary = (
    df_salary
    .groupby('formatted_experience_level')['normalized_salary']
    .median()
    .sort_values()
)

fig, ax = plt.subplots(figsize=(10, 5))
exp_salary.plot.bar(ax=ax, color='steelblue', edgecolor='white')
ax.set_title('Median Salary by Experience Level', fontsize=14)
ax.set_xlabel('Experience Level')
ax.set_ylabel('Median Salary (USD)')
ax.tick_params(axis='x', rotation=30)
plt.tight_layout()
plt.show()
```

```python
# Top 10 locations ÔÇö horizontal bar (most postings at top)
top_locations = df_fulltime['location'].value_counts().head(10).sort_values()

fig, ax = plt.subplots(figsize=(10, 6))
top_locations.plot.barh(ax=ax, color='steelblue', edgecolor='white')
ax.set_title('Top 10 Locations ÔÇö Full-Time Postings', fontsize=14)
ax.set_xlabel('Number of Postings')
ax.set_ylabel('Location')
plt.tight_layout()
plt.show()
```

## ­ƒîê Seaborn ÔÇö Heatmaps and Distributions <a class="anchor" id="seaborn"></a>

### ­ƒº® 4. Heatmap ÔÇö Two Categorical Dimensions

Heatmaps show how a metric varies across two categorical axes ÔÇö colour encodes the value.

```python
# Posting count by experience level ├ù work type
pivot = (
    df_postings
    .groupby(['formatted_experience_level', 'formatted_work_type'])
    .size()
    .unstack(fill_value=0)
)
pivot
```

```python
fig, ax = plt.subplots(figsize=(12, 5))

sns.heatmap(pivot, cmap='Blues', annot=True, fmt='d', ax=ax)

ax.set_title('Posting Count by Experience Level ├ù Work Type', fontsize=14)
plt.tight_layout()
plt.show()
```

### ­ƒº® 5. Histogram ÔÇö Distribution of a Numeric Column

Histograms show how values are spread ÔÇö where most values cluster, and whether there are outliers or a long tail.

```python
# Salary distribution with KDE and median line
fig, ax = plt.subplots(figsize=(12, 5))

sns.histplot(df_salary['normalized_salary'], bins=100, kde=True, ax=ax, color='steelblue')
median_val = df_salary['normalized_salary'].median()

ax.axvline(median_val, color='coral', linewidth=2, linestyle='--', label=f'Median: ${median_val:,.0f}')
ax.set_title('Distribution of Normalized Salary', fontsize=14)
ax.set_xlabel('Annual Salary (USD)')
ax.set_ylabel('Count')
ax.legend()
plt.tight_layout()
plt.show()
```

### ­ƒº® 6. Boxplot ÔÇö Distribution by Category

A boxplot shows the median, quartiles, and outliers for each group ÔÇö more information than a bar chart.

```python
# Salary distribution by experience level ÔÇö ordered by median
order = (
    df_salary
    .groupby('formatted_experience_level')['normalized_salary']
    .median()
    .sort_values()
    .index
)

fig, ax = plt.subplots(figsize=(12, 5))

sns.boxplot(data=df_salary, x='formatted_experience_level', y='normalized_salary',
            hue='formatted_experience_level', order=order, ax=ax,
            palette='Blues', legend=False)

ax.set_title('Salary Distribution by Experience Level', fontsize=14)
ax.set_xlabel('Experience Level')
ax.set_ylabel('Annual Salary (USD)')
ax.set_yscale('log')
ax.tick_params(axis='x', rotation=30)
plt.tight_layout()
plt.show()
```

## ÔÜí Plotly Express ÔÇö Interactive Charts (Optional) <a class="anchor" id="plotly"></a>

Plotly charts are interactive ÔÇö hover to see values, click to filter, zoom in. Useful for exploration.

Install if needed: `pip install plotly`

```python
import plotly.express as px

# Interactive bar chart ÔÇö median salary by experience level
exp_salary_df = (
    df_salary
    .groupby('formatted_experience_level')['normalized_salary']
    .agg(['median','count'])
    .reset_index()
    .sort_values('median')
)

fig = px.bar(
    exp_salary_df,
    x='formatted_experience_level',
    y='median',
    hover_data=['count'],
    title='Median Salary by Experience Level',
    labels={'formatted_experience_level': 'Experience Level', 'median': 'Median Salary (USD)', 'count': 'Postings with Salary'}
)
fig.show()
```

```python
# Interactive histogram ÔÇö salary distribution
fig = px.histogram(
    df_salary,
    x='normalized_salary',
    nbins=60,
    title='Salary Distribution',
    labels={'normalized_salary': 'Annual Salary (USD)'}
)
fig.show()
```
