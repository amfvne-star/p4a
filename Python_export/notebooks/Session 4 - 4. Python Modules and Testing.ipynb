## ­ƒôª What is a Python Module? <a class="anchor" id="module"></a>

So far, all your code has lived inside Jupyter notebooks ÔÇö cells that mix code, outputs, and text.

A **Python module** is simply a `.py` file that contains Python code (usually functions). You import it into a notebook or another script and call its functions.

| | Notebook (`.ipynb`) | Module (`.py`) |
|---|---|---|
| **Purpose** | Exploration, narrative, outputs | Reusable logic |
| **Runs** | Cell by cell, interactively | Imported, not run directly |
| **Testable** | Not directly | Yes ÔÇö with pytest |
| **Readable** | Great for humans | Great for other code |

**The pattern used in professional data science teams:**
- Write the logic in a `.py` module
- Import it into the notebook for the narrative and outputs
- Test the logic with pytest

Your notebook stays clean. Your logic stays testable.

### ­ƒº® 3. The problem ÔÇö and the fix

By default, Python caches imported modules. If you edit `analysis.py` and re-run the import cell, the notebook still uses the old version.

**The fix:** load the `autoreload` extension at the top of your notebook. It detects changes and reloads the module automatically.

````python
# Run this once at the top of any notebook that imports a .py module
%load_ext autoreload
%autoreload 2

# After this, any change you save to analysis.py is picked up automatically
# without needing to restart the kernel
````

## ­ƒôØ Docstrings and Type Hints <a class="anchor" id="docstrings"></a>

### ­ƒº® 4. Making your functions self-documenting

A **docstring** is a string written immediately after the `def` line. It explains what the function does. Type hints tell you what type each argument and return value should be.

```python
from sqlalchemy import Engine

def load_postings(engine: Engine) -> pd.DataFrame:
    """
    Load the postings table from the MySQL database.
    Parses listed_time to a UTC-aware datetime.

    Parameters
    ----------
    engine : sqlalchemy.Engine
        Active SQLAlchemy connection to the course database.

    Returns
    -------
    pd.DataFrame
        DataFrame with a parsed listed_time column (UTC).
    """
    df = pd.read_sql("SELECT * FROM postings", engine)
    df['listed_time'] = pd.to_datetime(df['listed_time'], utc=True)
    return df
```

You do not need to memorise this format ÔÇö but you should be able to read it. When you see `engine: Engine`, it means the argument should be a SQLAlchemy engine. When you see `-> pd.DataFrame`, it means the function returns a DataFrame.

In JupyterLab, `Shift+Tab` inside a function call shows the docstring ÔÇö try it on any Pandas function.

````python
# See the docstring of a function you already know
import pandas as pd
help(pd.read_sql)  # or place cursor inside pd.read_sql() and press Shift+Tab
````
